#Embedding层
Embedding主要解决离散数据如何进入神经网络的问题。

````python
class torch.nn.Embedding(num_embeddings, embedding_dim, padding_idx=None, max_norm=None, norm_type=2.0,
            scale_grad_by_freq=False, sparse=False, _weight=None, _freeze=False, device=None, dtype=None)
#num_embeddings：词表总数，即不同的ID总数。
#embedding_dim：每个词映射成多少维的向量
#padding_idx：由于不同句子长度不同，为了组成Batch,需要padding，默认为0，即用第0个此填充，PyTorch 会让 padding 对应的位置不会正常参与梯度更新。
#max_norm：进行归一化，限制Embedding向量的大小
#scale_grad_by_freq：一个 ID 出现得越频繁，它的梯度会按照出现次数进行缩放。
#sparse：让梯度使用：sparse gradient，避免生成完整巨型梯度矩阵。只在少数情况下支持使用，一般都不用。
````

Embedding 本质上就是一个：``可训练查找表``

假设词表：
$$
V=10000
$$Embedding 维度：
$$
D=256
$$
那么：


```
nn.Embedding(
    num_embeddings=10000,
    embedding_dim=256
)
```
里面实际上有一个参数矩阵：
$$
E\in R^{10000\times256}
$$可以理解成：
```
Token 0 → E[0]
Token 1 → E[1]
Token 2 → E[2]
...
Token 9999 → E[9999]
```
每一个 Token 对应一个256维向量。

##Embedding需要训练
刚开始训练时每个token各自对应初始随机向量，没有任何语义，随着反向传播不断训练，有些词可能变得比较接近，有些词可能变得比较远，比如cat dog学习得比较近，fish car学习得比较远，即模型会自动学习离散对象之间的语义关系。

前向时，Embedding就是完成纯粹的索引查表工作，将离散数据转化为输入张量。

Embedding是Transformer输入端最基础的组件之一。
