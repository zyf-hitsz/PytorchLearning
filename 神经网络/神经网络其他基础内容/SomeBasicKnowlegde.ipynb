#神经网络基础结构

##多层感知机

####什么是感知机？
我们希望根据输入序列，对输入进行判断、识别等，感知机做的事情非常简单：
$$
z=\mathbf{w}^{T}\mathbf{x}+b
$$然后根据 \(z\) 的正负进行判断：
$$
\hat y=
\begin{cases}
1,& z\ge 0\\
0,& z<0
\end{cases}
$$
完整过程就是：
$$
\mathbf{x}
\rightarrow
\mathbf w^T\mathbf x+b
\rightarrow
\text{threshold}
\rightarrow
\hat y
$$这就是一个最基本的感知机。

感知机和神经元的结构几乎就是相同的，只不过感知机使用的非线性函数是阶跃函数。感知机的几何意义是寻找一个超平面，把两类数据分开，判断点位于超平面的哪一侧。单层感知机的超平面是线性的，只能解决线性可分问题，于是产生了多层感知机Multi-Layer Perceptron，即MLP。

####什么是多层感知机？
最经典的MLP：
```
Input Layer
    ↓
Linear
    ↓
Activation
    ↓
Linear
    ↓
Activation
    ↓
...
    ↓
Linear
    ↓
Output
```
$$
f(\mathbf x)
=
W_L
\sigma(
W_{L-1}
\sigma(
\cdots
\sigma(W_1\mathbf x+b_1)
))
+b_L
$$这就是 MLP 的核心数学形式。

MLP通常由输入层、隐藏层和输出层组成，下面是一个典型的MLP结构：
```
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Flatten(),

    nn.Linear(784, 256),
    nn.ReLU(),

    nn.Linear(256, 128),
    nn.ReLU(),

    nn.Linear(128, 10)
)
```
激活函数是 MLP真正拥有表达能力的关键,有了激活函数，MLP才有非线性表达能力。


####几个概念
深度：表示MLP有多少层变换。

宽度：每层有多少个神经元。



####多输入多输出感知机
一个 neuron：
$$
x\in\mathbb R^n
$$输出一个数：
$$
y\in\mathbb R
$$如果有：
$
m
$个 neuron：
$$
y_1=w_1^Tx+b_1
$$
$$
y_2=w_2^Tx+b_2
$$...
$$
y_m=w_m^Tx+b_m
$$把所有权重堆起来：
$$
W=
\begin{bmatrix}
w_1^T\\
w_2^T\\
\vdots\\
w_m^T
\end{bmatrix}
$$于是：
$$
y=Wx+b
$$

```
MLP的优点：
- 结构简单
- 能够逼近非常复杂的非线性函数
- 输入输出尺寸灵活
- 应用面广
```
```
MLP的主要缺点：
没有利用数据的先验结构
```
