#Flatten层
作用是把多为特征张量“摊平”成一维特征向量，方便送入全连接层。
例如 CNN 中卷积之后得到：
$$
x.shape = [N, C, H, W]
$$假设：
$$
[N,C,H,W]=[32,64,7,7]
$$经过 Flatten：
$$
[32,64,7,7]
\rightarrow
[32,64\times7\times7]
=
[32,3136]
$$这里 32 是 batch size，不会被展开掉，只把每个样本内部的 C、H、W 合并起来。

```python
import torch
import torch.nn as nn

# 假设卷积层输出：[batch_size, channels, height, width]
x = torch.randn(3, 3, 2, 2)

flatten = nn.Flatten(start_dim=1)
y = flatten(x)

print("Flatten前：", x.shape)
print(x)
print("Flatten后：", y.shape)
print(y)
```
